```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the CSV datasheet
df = pd.read_csv(r"C:\Users\TEST\Downloads\python.xlsx")
# Display the datasheet 
print("Datasheet")
print(df)
```

```python
# Line plot
plt.figure(figsize=(7, 5))
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks for Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
# Bar Plot
plt.figure(figsize=(7, 5))
plt.bar(df["Student"], df["Maths"])
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.grid(True)
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(7, 5))
plt.scatter(df["Maths"], df["Python"], s=70)
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
# Correlation Matrix
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("Correlation Matrix:")
print(correlation)
```

```python
# Heatmap
plt.figure(figsize=(7, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# Pie chart
y = df["Maths"]
plt.pie(y, labels = df["Student"])
plt.title("Maths Marks of Students")
plt.show()
```

```python
# Histogram
plt.figure(figsize=(7, 5))
plt.hist(df['Attendance'], bins=3, edgecolor='black')
plt.xlabel('Percentage')
plt.ylabel('Frequency')
plt.title('Histogram of Attendance of Students')
plt.show()
```
