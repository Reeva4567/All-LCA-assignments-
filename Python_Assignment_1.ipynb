```python
input_string = input("Enter a string: ")
frequency = {}

for letter in input_string:
    frequency[letter] = frequency.get(letter, 0) + 1

print("Letter frequencies:", frequency)
```

```python
numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

# divmod returns tuple (quotient, remainder)
result = divmod(numerator, denominator)

print("Output (Quotient, Remainder):", result)
```

```python
from datetime import date

def get_date_tuple(prompt_name):
    print(f"\nEnter details for {prompt_name}:")
    d = int(input("Enter day (1-31): "))
    m = int(input("Enter month (1-12): "))
    y = int(input("Enter year (e.g., 2026): "))
    return (d, m, y)

# Input date tuples
date_tuple1 = get_date_tuple("Date 1")
date_tuple2 = get_date_tuple("Date 2")

# Create date objects (Year, Month, Day)
d1 = date(date_tuple1[2], date_tuple1[1], date_tuple1[0])
d2 = date(date_tuple2[2], date_tuple2[1], date_tuple2[0])

# Calculate difference in days
difference = abs((d2 - d1).days)

print(f"\nNumber of days between the two dates: {difference} days")
```
