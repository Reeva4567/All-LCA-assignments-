```python
import math
def right_triangle(a, b, c):
    # Determine the largest side (hypotenuse) and the other two sides
    if a >= b and a >= c:
        largest = a
        x, y = b, c
    elif b >= a and b >= c:
        largest = b
        x, y = a, c
    else:
        largest = c
        x, y = a, b

    # Check Pythagoras theorem: largest^2 == x^2 + y^2
    if math.isclose(largest**2, x**2 + y**2):
        print("The triangle is right-angled.")
    else:
        print("The triangle is not right-angled.")

# User Input
a = float(input("Enter first side: "))
b = float(input("Enter second side: "))
c = float(input("Enter third side: "))

# Function Call
right_triangle(a, b, c)
```
